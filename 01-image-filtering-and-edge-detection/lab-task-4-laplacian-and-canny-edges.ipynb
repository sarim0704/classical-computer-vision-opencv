```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load grayscale image
img = cv2.imread('/content/Il.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError("Check the image path!")

# Compute Laplacian of original image
laplacian_original = cv2.Laplacian(
    img, cv2.CV_64F
)

# Convert to absolute values for display
laplacian_original = cv2.convertScaleAbs(
    laplacian_original
)

# Apply Gaussian blur (5x5, sigma=1.5)
blurred = cv2.GaussianBlur(
    img, (5, 5), 1.5
)

# Compute Laplacian after blurring
laplacian_blurred = cv2.Laplacian(
    blurred, cv2.CV_64F
)

laplacian_blurred = cv2.convertScaleAbs(
    laplacian_blurred
)

# Display results side by side
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(laplacian_original, cmap='gray')
plt.title('Original Image - Laplacian')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(laplacian_blurred, cmap='gray')
plt.title('Gaussian Blur + Laplacian')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Load grayscale image
img = cv2.imread('/content/Il.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError("Check the image path!")

# Apply Canny with three threshold pairs
edges_1 = cv2.Canny(img, 50, 100)
edges_2 = cv2.Canny(img, 100, 200)
edges_3 = cv2.Canny(img, 150, 300)

# Display all three edge maps
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(edges_1, cmap='gray')
plt.title('Canny: Thresholds (50, 100)')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(edges_2, cmap='gray')
plt.title('Canny: Thresholds (100, 200)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(edges_3, cmap='gray')
plt.title('Canny: Thresholds (150, 300)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
